Завдання 3. Статистичний аналіз досвіду респондентів



```python
mean = df['WorkExp'].mean()
median = df['WorkExp'].median()
mode = df['WorkExp'].mode()[0]
print(f"Measures of Central Tendency for Work Experience:")
print(f"Mean: {mean:.2f}")
print(f"Median: {median}")
print(f"Mode: {mode}")
```

Завдання 5. Визначення популярності Python



```python
df['worked_with_python'] = df['LanguageHaveWorkedWith'].str.contains('python', case=False, na=False)
percent_python = (df['worked_with_python'].sum() / df.shape[0]) * 100
print(f"Percentage of respondents who program in Python: {percent_python:.2f}%")
```

Завдання 6. Аналіз шляхів навчання програмуванню


```python
df['learned_with_online_courses'] = df['LearnCode'].str.contains('Online Courses', case=False, na=False)
online_count = df['learned_with_online_courses'].sum()
print(f"Number of respondents who learned with online coursess : {online_count:}")
```

Завдання 7. Географічний аналіз компенсації Python-розробників

```python
worked_python = df[df['worked_with_python']].dropna(subset=['ConvertedCompYearly'])
worked_by_country = worked_python.groupby('Country')['ConvertedCompYearly'].agg(['mean','median']).round(1)
worked_by_country.head()
```

Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів



```python
top_paid = df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']].dropna(subset=['ConvertedCompYearly'])
result = top_paid.sort_values(by='ConvertedCompYearly', ascending=False).reset_index(drop=True).head(5)
result
```

Завдання 9. Аналіз популярності Python по віковим категоріям

```python
age_python_stats = (df.groupby('Age')['worked_with_python'].mean()* 100).round(2)
age_python_stats
```

Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників

```python
top_75 = df['ConvertedCompYearly'].quantile(0.75)
high_paid_remote = df[(df['RemoteWork'] == 'Remote') & (df['ConvertedCompYearly'] > top_75)]
result = high_paid_remote['Industry'].dropna().value_counts()
result
```
